This function introduces the rule of the game

```python
def intro():
    print("Welcome to Tic Tac Toe! by Thanh")
    print("\n")
    print("Rules: Player 1 and Player 2 chose their marks (X or O)" 
          "and take turns placing them on the 3x3 grid. " \
          "The player who succeeds in placing three of " \
          "their marks in a horizontal, vertical, or diagonal" \
          "row wins the game.")
    input("Press Enter to start the game...")
    print("Let's start the game!")
```

This function creates a board

```python
def create_grid():
    print("here is the grid")
    grid = [[" " for _ in range(3)] for _ in range(3)]
    return grid
```

This function define the players symbols

```python
def symbols():
    symbol_1 = input("Player 1, please choose your symbol (X or O): ")
    while symbol_1 not in ["X", "O"]:
        print("Invalid input. Please choose either X or O.")
        symbol_1 = input("Player 1, please choose your symbol (X or O): ")
        if symbol_1 == "X":
            symbol_2 = "O"
            print("Player 2, your symbol is O.")
        else:
            symbol_2 = "X"
            print("Player 2, your symbol is X.")
    input("Press Enter to continue...")
    return symbol_1, symbol_2
```
